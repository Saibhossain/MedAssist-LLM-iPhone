````python
from unsloth import FastLanguageModel
import torch

max_seq_length = 1024

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Qwen2.5-0.5B-Instruct",
    max_seq_length = max_seq_length,
    dtype = torch.float16,
    load_in_4bit = False,   # keep full precision during training
)
````

````python
model = FastLanguageModel.get_peft_model(
    model,
    r = 16,                       # rank of adapter matrices
    target_modules = [
        "q_proj","k_proj","v_proj","o_proj",
        "gate_proj","up_proj","down_proj",
    ],
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",
    use_gradient_checkpointing = True,  # saves memory
)
````

````python
model.print_trainable_parameters()
````

````python
# load Medical dataset
from datasets import load_dataset
dataset = load_dataset("lavita/MedQuAD", split="train[:3000]")
print(dataset)
print(dataset[0])
````

````python
def format_med(x):
    return {
        "text": f"""### Instruction:
Answer the medical question clearly and say this is not a diagnosis.

### Question:
{x['question']}

### Response:
{x['answer']}"""
    }

dataset = dataset.map(format_med)

print(dataset[0]["text"][:400])
````

````python
print(dataset.column_names)
print(dataset[0]["text"][:300])
````

````python
def tokenize(batch):
    tokens = tokenizer(
        batch["text"],
        truncation=True,
        padding="max_length",
        max_length=512,
    )
    tokens["labels"] = tokens["input_ids"].copy()
    return tokens

dataset = dataset.map(tokenize, batched=True)
````

````python
sample = dataset[0]

print("Token IDs:", sample["input_ids"][:30])
print("Decoded:")
print(tokenizer.decode(sample["input_ids"][:200]))
````

````python
dataset = dataset.remove_columns(
    [c for c in dataset.column_names if c not in ["input_ids","attention_mask","labels"]]
)

print(dataset)
````

````python
import torch

ids = torch.tensor(dataset[0]["input_ids"])
labels = torch.tensor(dataset[0]["labels"])

print("Match:", torch.equal(ids, labels))
````

````python
from transformers import TrainingArguments

training_args = TrainingArguments(
    per_device_train_batch_size = 2,
    gradient_accumulation_steps = 4,
    num_train_epochs = 1,
    learning_rate = 2e-4,
    fp16 = True,
    logging_steps = 5,
    save_steps = 50,
    output_dir = "debug_outputs",
    report_to = "none",
)
````

````python
from transformers import Trainer

class DebugTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        outputs = model(**inputs)
        loss = outputs.loss

        print("\n=== DEBUG STEP ===")
        print("Input shape:", inputs["input_ids"].shape)
        print("Loss:", float(loss))

        return (loss, outputs) if return_outputs else loss
````

````python
trainer = DebugTrainer(
    model = model,
    args = training_args,
    train_dataset = dataset,
)
````

````python
batch = {k: torch.tensor([dataset[0][k]]).to(model.device)
         for k in ["input_ids","attention_mask","labels"]}

with torch.no_grad():
    out = model(**batch)

print("Pre-train loss:", out.loss.item())
````

````python
trainer.train()
````

````python
for name, param in model.named_parameters():
    if param.requires_grad and param.grad is not None:
        print(name, param.grad.abs().mean().item())
        break
````

````python
model = model.merge_and_unload()

model.save_pretrained("medicalQA_finetuned_model")
tokenizer.save_pretrained("medical_finetuned_model")
````

````python
# Reload Clean Model

from unsloth import FastLanguageModel # Import FastLanguageModel
import torch

# Use FastLanguageModel to load the merged model compatible with unsloth's patches
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="medical_mobile_ready",
    max_seq_length=1024, # Use the previously defined max_seq_length
    dtype=torch.float16,
    load_in_4bit=False, # Load in full precision as it's already merged
    device_map="auto"
)

model.eval()
print("Reloaded OK")
````

````python
# Sanity Test (Medical Behavior Check)

prompt = """### Instruction:
Answer clearly but not as diagnosis.

### Question:
What is diabetes?

### Response:
"""

inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

out = model.generate(**inputs, max_new_tokens=120)
print(tokenizer.decode(out[0]))
````

````python
# Medical Eval Loop
tests = [
    "What is hypertension?",
    "What causes asthma?",
    "What is anemia?",
    "What is stroke?",
]

for q in tests:
    prompt = f"""### Instruction:
Answer clearly but not as diagnosis.

### Question:
{q}

### Response:
"""
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    out = model.generate(**inputs, max_new_tokens=120)
    print("\n---")
    print(tokenizer.decode(out[0]))
````

# Mobile Deployment Pipeline

````python
#Uninstall xformers to prevent conflicts with ExecuTorch tracing
!pip uninstall xformers -y

#Reload Clean Model

from unsloth import FastLanguageModel
import torch

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="medical_mobile_ready",
    max_seq_length=1024,
    dtype=torch.float16,
    load_in_4bit=False,
    device_map="cpu",   # export works best from CPU
)

model.eval()
print("Model ready for export")

FastLanguageModel.for_inference(model)

# Explicitly set attention implementation to eager *after* for_inference and before quantization/capture
model.config.attn_implementation = "eager"

import torch

quantized_model = torch.quantization.quantize_dynamic(
    model,
    {torch.nn.Linear},
    dtype=torch.qint8
)

print("Quantized to INT8")

example = tokenizer(
    "What is diabetes?",
    return_tensors="pt"
)

example_ids = example["input_ids"]

import executorch.exir as exir

captured = exir.capture(
    quantized_model,
    (example_ids,), # Wrap example_ids in a tuple
)
et_program = captured.to_executorch()

with open("medical_llm_mobile.pte", "wb") as f:
    f.write(et_program.buffer)

print("Saved mobile model")
with open("medical_llm_mobile.pte", "wb") as f:
    f.write(et_program.buffer)

print("Saved mobile model")
````
